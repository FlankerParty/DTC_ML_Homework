Pandas version: 3.0.6

```python
import pandas as pd
source_data = pd.read_csv("car_fuel_efficiency_2026.csv")
print(f"Pandas version: {pd.__version__}")
```

Records count: 10000

```python
row_cnt = source_data.shape[0]
print(f"Records count: {row_cnt}")
```

Fuel Types: 3

```python
fuel_types_cnt = source_data.drop_duplicates(subset=["fuel_type"])["fuel_type"].count()
print(f"Fuel Types: {fuel_types_cnt}")
```

Columns with missing value: 2

```python
missing_values_cnt = (source_data.isna().any(axis=0).sum())
print(f"Columns with missing value: {missing_values_cnt}")
```

Max fuel efficiency of cars in Asia: 41.2

```python
max_fuel_efficiency_Asia = source_data[source_data["origin"] == "Asia"]["fuel_efficiency_mpg"].max()
print(f"Max fuel efficiency of cars in Asia: {max_fuel_efficiency_Asia}")
```

```python
mid_horsepower_old = source_data["horsepower"].median()
mode_val = source_data["horsepower"].mode().iloc[0]
source_data_new = source_data
source_data_new["horsepower"] = source_data_new["horsepower"].fillna(mode_val)
mid_horsepower_new = source_data_new["horsepower"].median()
print(f"Midian value of horsepower in old dataset: {mid_horsepower_old}")
print(f"Midian value of horsepower in new dataset: {mid_horsepower_new}")
```

Sum of elements in w: 0.36919696904927707

```python
import numpy as np
tmp_dataset = source_data[source_data["origin"] == "Asia"][["vehicle_weight","model_year"]].head(7)
X = tmp_dataset.values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
print(f"Sum of elements in w: {w.sum()}")
```
